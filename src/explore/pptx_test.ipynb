```python
from pptx import Presentation
```

```python
def extract_text_from_pptx(file_path:str):

    text = []

    presentation = Presentation(file_path)
    for slides in presentation.slides:
        print(f"Slide {slides.slide_id}:")

        for shape in slides.shapes:
            if shape.has_text_frame:
                # text.append(shape.text)
                print(f"Text: {shape.text}")
                for paragraph in shape.text_frame.paragraphs:
                    text.append(paragraph.text)
            elif shape.has_table:
                for row in shape.table.rows:
                    for cell in row.cells:
                        text.append(cell.text)
            elif shape.has_chart:
                chart = shape.chart
                for series in chart.series:
                    for point in series.points:
                        try:
                            label = point.data_label.text_frame.text
                            if label:
                                text.append(str(label))
                        except(AttributeError, KeyError):
                            pass


    return text
```

```python
file_path = "/Users/rohit/Documents/Git/Enterprise_RAG_Pipeline/data/raw/pharma_complex_dashboards_set.pptx"
extracted_text = extract_text_from_pptx(file_path)
for context in extracted_text:
    print(context)
```
